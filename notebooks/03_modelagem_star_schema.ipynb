### Conexão com banco

```python
import os
import urllib
import prettytable  # Importado para corrigir o bug de versão
from dotenv import load_dotenv
```

```python
if 'DEFAULT' not in prettytable.__dict__:
    try:
        from prettytable import Style
        prettytable.__dict__['DEFAULT'] = Style.DEFAULT
    except ImportError:
        prettytable.__dict__['DEFAULT'] = 10  # ID histórico do estilo padrão

# 1. Carrega as variáveis de ambiente com segurança do arquivo .env
load_dotenv(os.path.join("..", ".env"))

SERVER = os.getenv('DB_SERVER')
DATABASE = "banco_teste"  # Seu banco de testes provisório
USER = os.getenv('DB_USER')
PASSWORD = os.getenv('DB_PASSWORD')
DRIVER = os.getenv('DB_DRIVER', 'ODBC Driver 17 for SQL Server')

# 2. Monta os parâmetros nativos do ODBC e faz o URL-Encode
odbc_str = (
    f"DRIVER={DRIVER};"
    f"SERVER={SERVER};"
    f"DATABASE={DATABASE};"
    f"UID={USER};"
    f"PWD={PASSWORD};"
    f"Encrypt=yes;"
    f"TrustServerCertificate=yes;"
)
connection_url = f"mssql+pyodbc:///?odbc_connect={urllib.parse.quote_plus(odbc_str)}"
```

```python
# 3. Limpa e reinicia a extensão de SQL na memória do Jupyter
%reload_ext sql

# 4. Conecta passando a string perfeitamente higienizada
%sql $connection_url

# 5. Configurações de exibição limpa
%config SqlMagic.feedback = False
%config SqlMagic.autopandas = True

print(f"🔌 Extensão %%sql conectada e corrigida com sucesso ao banco [{DATABASE}]!")
```
